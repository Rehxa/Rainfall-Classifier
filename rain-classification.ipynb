### **Missing Values Handling**

### **Outliers Handling**

```python
print("\n" + "="*60)
print("OUTLIER DETECTION & HANDLING")
print("="*60)

# Identify numeric columns (excluding date + rain)
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
numeric_cols = [col for col in numeric_cols if col not in ['date', 'rain', 'precipitation', 'weather_code']]

# Define physical limits for known weather variables
physical_limits = {
    'relative_humidity_2m': (0, 100),
    'humidity': (0, 100),
    'cloud_cover': (0, 100),
    'surface_pressure': (950, 1050),
    'pressure': (950, 1050),
    'temperature_2m': (-30, 50),
    'temperature': (-30, 50)
}

# Store outlier stats
outlier_stats = []

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower, upper = Q1 - 1.5*IQR, Q3 + 1.5*IQR
    
    # Count outliers
    outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    if outliers > 0:
        outlier_stats.append({'column': col, 'outliers': outliers, 'percentage': outliers/len(df)*100})
    
    # Apply physical limits if defined
    if col in physical_limits:
        df[col] = df[col].clip(*physical_limits[col])
    else:
        df[col] = df[col].clip(lower, upper)

# Summary
print(f"\nDetected outliers in {len(outlier_stats)} columns:")
for stat in outlier_stats:
    print(f" - {stat['column']}: {stat['outliers']} ({stat['percentage']:.1f}%)")
print("\n✓ Outlier handling complete!")
```

### **Convert Date to Useful Features**

```python
df['date'] = pd.to_datetime(df['date'])

df['hour'] = df['date'].dt.hour
df['day'] = df['date'].dt.day
df['month'] = df['date'].dt.month
df['day_of_week'] = df['date'].dt.dayofweek

# Optional: drop original date
df.drop('date', axis=1, inplace=True)
```

### **Create Target Variable (Rain Classification)**

```python
def classify_rain(x):
    if x == 0:     return 0 # No rain
    elif x <= 0.2: return 1 # Light rain
    else:          return 2 # Heavy rain

df['rain_class'] = df['rain'].apply(classify_rain)
```

drop highly irrelevant features

### **Create Rolling / Lag Features (For Time Series)**

```python
#Purpose:

#Smooth out short-term fluctuations
#Highlight longer-term trends
#Reduce the “noise” in the data

rolling_cols = ['temperature_2m', 'relative_humidity_2m', 'surface_pressure']

for col in rolling_cols:
    # 3-hour rolling average
    df[f'{col}_roll3'] = df[col].rolling(window=3).mean().bfill()
    
    # 6-hour rolling average
    df[f'{col}_roll6'] = df[col].rolling(window=6).mean().bfill()
```

### **Feature Interactions**

```python
# combined features help the model see relationships more clearly

df['temp_humidity'] = df['temperature_2m'] * df['relative_humidity_2m']
df['pressure_temp'] = df['surface_pressure'] * df['temperature_2m']
df['temp_diff'] = df['temperature_2m'] - df['dew_point_2m']  

df.dtypes
# df.shape
```

### **Feature Transformation / Scaling**

```python
from sklearn.preprocessing import MinMaxScaler

# Get all feature columns except the target label
features_to_scale = [col for col in df.columns if col != 'rain_class']

# Fit the scaler and transform all features to range [0, 1]
scaler = MinMaxScaler()
df[features_to_scale] = scaler.fit_transform(df[features_to_scale])
print("Scaling complete!")
df[features_to_scale]
```
